# Final Coupled Flood–Earthquake Equations in 4D Spacetime (Two Time Variables)

## 1. Spacetime & Geometric Algebra Setup
- **Coordinates**: $X = (t_1, t_2, x_1, x_2)$  
- **Clifford algebra**: $\mathcal{G}_{2,2}$ with basis $\{\gamma_{t_1}, \gamma_{t_2}, \gamma_{x_1}, \gamma_{x_2}\}$  
- **Metric**: $\gamma_a \cdot \gamma_b = \eta_{ab} = \mathrm{diag}(+1, +1, -1, -1)$  
- **Spacetime derivative**: $\nabla = \gamma^a \partial_a$, where $\partial_a = \frac{\partial}{\partial X^a}$  
- **Multivector fields**:  
  - $M$ (flood momentum) → vector in $\mathrm{span}\{\gamma_{t_1}, \gamma_{x_1}\}$  
  - $\lambda$ (Lagrange multiplier) → scalar  
  - $\mathbf{u}$ (earthquake displacement) → vector in $\mathrm{span}\{\gamma_{t_2}, \gamma_{x_2}\}$  

---

## 2. Total Action Functional
$$
I_{\text{total}}[M, \lambda, \mathbf{u}] = 
\int_{\mathcal{M}} 
\Big[ 
\langle \lambda \, (\nabla M - S[\mathbf{u}]) \rangle_0 
+ 
\underbrace{\frac{1}{2}\rho \dot{\mathbf{u}}^2 - \frac{1}{2}(\lambda+2\mu)(\nabla \cdot \mathbf{u})^2 + \frac{1}{2}\mu(\nabla \wedge \mathbf{u})^2}_{\mathcal{L}_{\text{seismic}}}
\Big] \, d^4X
$$
with the coupling source:
$$
S[\mathbf{u}] = -g h \, \frac{\partial z_b(\mathbf{u})}{\partial x_1} \, \gamma_{t_1}
$$

---

## 3. Final Coupled Euler–Lagrange Equations (GA Form)
Varying $I_{\text{total}}$ w.r.t. $\lambda$, $M$, and $\mathbf{u}$ yields the **core 3‑field system**:

$$
\boxed{
\begin{aligned}
&\text{(i) Flood Conservation:} && \nabla M = S[\mathbf{u}] \\[6pt]
&\text{(ii) Adjoint Constraint:} && \tilde{\nabla}\tilde{\lambda} = 0 \\[6pt]
&\text{(iii) Seismic Wave Eq.:} && -\rho \ddot{\mathbf{u}} + (\lambda+2\mu)\nabla(\nabla \cdot \mathbf{u}) + \mu \nabla \cdot (\nabla \wedge \mathbf{u}) = \mathbf{F}_{\text{flood}}
\end{aligned}
}
$$
where the flood forcing on the earthquake is:
$$
\boxed{
\mathbf{F}_{\text{flood}} = \frac{\partial}{\partial \mathbf{u}} \langle \lambda \, S[\mathbf{u}] \rangle_0
}
$$

---

## 4. Expanded Physical Component Form (with two times)
Expanding the GA equations into PDEs using $t_1, t_2, x_1, x_2$:

**(a) Flood Equations** (evolve along $t_1$):
$$
\boxed{
\begin{aligned}
&\partial_{t_1} h + \partial_{x_1} (h u) = 0 \\[6pt]
&\partial_{t_1} (h u) + \partial_{x_1}\left(h u^2 + \frac{1}{2} g h^2\right) = -g h \, \frac{\partial z_b}{\partial x_1}
\end{aligned}
}
$$

**(b) Seismic Wave Equation** (evolves along $t_2$, forced by flood):
$$
\boxed{
\rho \, \ddot{\mathbf{u}} - (\lambda+2\mu)\nabla(\nabla \cdot \mathbf{u}) - \mu \nabla^2 \mathbf{u} 
= 
-g h \, \frac{\partial}{\partial \mathbf{u}}\left(\frac{\partial z_b}{\partial x_1}\right) \lambda
}
$$
where $\ddot{\mathbf{u}} = \partial_{t_1}^2 \mathbf{u} + \partial_{t_2}^2 \mathbf{u} + 2\partial_{t_1}\partial_{t_2}\mathbf{u}$.

**(c) Interface Coupling** (bed moves with ground motion):
$$
\boxed{
\frac{\partial z_b}{\partial t_2} = -\frac{\partial u_y}{\partial t_2}
}
$$
where $u_y$ is the vertical component of $\mathbf{u}$.

## 5. Wave‑Mode Decomposition (P & S waves with Flood Sources)
Decompose $\mathbf{u} = \nabla \phi + \nabla \wedge \mathbf{w}$. The final decoupled wave equations are:

$$
\boxed{
\begin{aligned}
&\text{P-wave:} && \nabla^2 \phi - \frac{1}{V_p^2}\ddot{\phi} = Q_p, 
& Q_p &= \frac{1}{\rho V_p^2} \langle \lambda \, \nabla \cdot S \rangle_0 \\[6pt]
&\text{S-wave:} && \nabla^2 \mathbf{w} - \frac{1}{V_s^2}\ddot{\mathbf{w}} = \mathbf{Q}_s, 
& \mathbf{Q}_s &= \frac{1}{\rho V_s^2} \langle \lambda \, \nabla \wedge S \rangle_0
\end{aligned}
}
$$

## 6. Compact Unified STA (Spacetime Algebra) Form
Define a unified multivector $\Psi = M + \mathbf{u} + \lambda$ and a unified wave operator $\Box = \nabla - \frac{1}{V}\partial_t$. The **entire system collapses to a single Dirac‑like equation**:

$$
\boxed{
\Box \, \Psi = \mathcal{J}_{\text{flood}} 
\quad \text{with} \quad
\mathcal{J}_{\text{flood}} = S[\mathbf{u}] + \mathbf{F}_{\text{flood}}
}
$$
subject to the subsidiary condition:
$$
\boxed{
\tilde{\nabla} \tilde{\lambda} = 0
}
$$

## 7. Angular (Rotational) Representation on the 6 Orthogonal Circles
Let each coordinate be an angle: $X^a \longrightarrow \theta^a$. The gradients become angular derivatives:
$$
\partial_a \longrightarrow \frac{1}{R_a} \frac{\partial}{\partial \theta^a}
$$
Thus, the final **angular‑form field equations** are:

$$
\boxed{
\begin{aligned}
&\frac{1}{R_{t_1}}\partial_{\theta_{t_1}} M + \frac{1}{R_{x_1}}\partial_{\theta_{x_1}} M = -g h \, \frac{1}{R_{x_1}}\partial_{\theta_{x_1}} z_b(\mathbf{u}) \, \gamma_{t_1} \\[6pt]
&\frac{1}{R_{t_2}}\partial_{\theta_{t_2}} z_b = -\frac{1}{R_{t_2}}\partial_{\theta_{t_2}} u_y \\[6pt]
&\rho \left( \frac{1}{R_{t_1}}\partial_{\theta_{t_1}} + \frac{1}{R_{t_2}}\partial_{\theta_{t_2}} \right)^2 \mathbf{u} 
- (\lambda+2\mu) \nabla_\theta(\nabla_\theta \cdot \mathbf{u}) 
- \mu \nabla_\theta^2 \mathbf{u} 
= -g h \, \frac{\partial}{\partial \mathbf{u}}\left( \frac{1}{R_{x_1}}\partial_{\theta_{x_1}} z_b \right) \lambda
\end{aligned}
}
$$

### 1. Discrete‑Time Dynamical System (Explicit Euler)

Let the state at step $k$ be $\mathbf{x}_k \in \mathbb{R}^n$, and let the step size be $h > 0$.  
Given a fixed permutation matrix $P$ (e.g., swapping coordinates $i$ and $j$) and a deterministic binary schedule $\{s_k\}_{k=0}^{\infty} \subset \{0,1\}$, the Euler iteration is:

$$
\mathbf{x}_{k+1} = \mathbf{x}_k + h \cdot \mathbf{F}^{(s_k)}(\mathbf{x}_k),
$$

where

$$
\mathbf{F}^{(s_k)}(\mathbf{x}_k) = \big[(1 - s_k) I + s_k P\big] \mathbf{F}(\mathbf{x}_k).
$$

Expanding the bracket:

$$
\boxed{\mathbf{x}_{k+1} = \mathbf{x}_k + h \Big( (1 - s_k)\,\mathbf{F}(\mathbf{x}_k) + s_k \, P\mathbf{F}(\mathbf{x}_k) \Big)}
$$

---

#### Explicit 2D Example
Let $n=2$, $\mathbf{x} = (x, y)^\top$, $\mathbf{F} = (F_x, F_y)^\top$, and $P$ be the full swap $\begin{pmatrix}0&1\\1&0\end{pmatrix}$.  
Then:

- If $s_k = 0$ (no swap):
  $$
  \begin{cases}
  x_{k+1} = x_k + h\, F_x(x_k, y_k) \\
  y_{k+1} = y_k + h\, F_y(x_k, y_k)
  \end{cases}
  $$

- If $s_k = 1$ (swap):
  $$
  \begin{cases}
  x_{k+1} = x_k + h\, F_y(x_k, y_k) \\
  y_{k+1} = y_k + h\, F_x(x_k, y_k)
  \end{cases}
  $$

---

### 2. General $n$-D Algorithm (Pseudo‑code)

For arbitrary dimension $n$ and a permutation $P$ (which may be a transposition or a longer cycle), the iteration is:

```text
Input: x_0 (initial state), F (function handle), P (n x n permutation matrix),
       h (step size), S = [s_0, s_1, ..., s_{N-1}] (user‑supplied binary array)
Output: trajectory x_0, x_1, ..., x_N

for k = 0 to N-1 do:
    F_val = F(x_k)
    if S[k] == 0:
        x_{k+1} = x_k + h * F_val
    else:
        x_{k+1} = x_k + h * (P @ F_val)   # matrix‑vector multiplication
end for
```

**Determinism guarantee:** Since $S$ is fixed *a priori* and no random number generator is used, every run with the same $S$, $h$, and $\mathbf{x}_0$ yields identical results.

---

### 3. Discrete Coupling Coefficient

Over $N$ iterations, the coupling coefficient exactly matches your empirical frequency definition:

$$
\boxed{C_N = \frac{1}{N} \sum_{k=0}^{N-1} s_k = \frac{\text{number of } 1\text{s in } S}{N}}
$$

This scalar lies in $[0,1]$. A schedule of all zeros gives $C_N = 0$ (uncoupled), all ones gives $C_N = 1$ (fully swapped at every step), and a mixture gives intermediate coupling strength.

---

### 4. Extending to Higher‑Order Deterministic Integrators (e.g., RK4)

Because the swap operator $L_{s} = (1-s)I + sP$ is a **linear operator** acting on the vector field, it commutes with scalar multiplication and addition. Therefore, you can apply it directly inside any Runge–Kutta stage without breaking determinism.

For RK4, at step $k$ with binary value $s = s_k$:

1. Compute the base field $\mathbf{F}_0 = \mathbf{F}(\mathbf{x}_k)$.
2. Compute the swapped field $\mathbf{F}_1 = L_s \mathbf{F}_0$.
3. Compute intermediate stages using the *same* linear operator:
   $$
   \begin{aligned}
   \mathbf{k}_1 &= L_s\, \mathbf{F}(\mathbf{x}_k), \\
   \mathbf{k}_2 &= L_s\, \mathbf{F}\!\left(\mathbf{x}_k + \frac{h}{2}\mathbf{k}_1\right), \\
   \mathbf{k}_3 &= L_s\, \mathbf{F}\!\left(\mathbf{x}_k + \frac{h}{2}\mathbf{k}_2\right), \\
   \mathbf{k}_4 &= L_s\, \mathbf{F}\!\left(\mathbf{x}_k + h\,\mathbf{k}_3\right),
   \end{aligned}
   $$
   $$
   \mathbf{x}_{k+1} = \mathbf{x}_k + \frac{h}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4).
   $$

This preserves the deterministic nature and maintains the same coupling operator throughout the substeps of a single macro‑iteration.

---

### 5. Constructing Deterministic Schedules

To remain strictly non‑stochastic, the user‑supplied array $S$ can be generated by any deterministic rule, such as an **arbitrary hard‑coded list**: $S = [0,1,1,0,1,0,\dots]$.

The choice of $S$ directly controls $C_N$, allowing you to dial the coupling strength deterministically.

---

### 6. Stability and Step‑Size Consideration

Since applying $P$ changes the direction of the vector field components, the Lipschitz constant of $\mathbf{F}^{(s)}$ may differ from that of $\mathbf{F}$. For explicit Euler, ensure the step size $h$ satisfies stability bounds for **both** the swapped and unswapped regimes, i.e.,

$$
h < \min_{s \in \{0,1\}} \frac{2}{\lambda_{\max}\left( J^{(s)} \right)},
$$

where $J^{(s)}$ is the Jacobian of $\mathbf{F}^{(s)}$, to maintain deterministic numerical stability without relying on adaptive random step sizes.

### The Geometry of 4D → 6 Angles

In 4 dimensions, you have 4 coordinate axes $(x_1, x_2, x_3, x_4)$.  
Between any two distinct axes, you can define a **relative angular relationship**—specifically, the angle  

$$
\theta_{ij} = \operatorname{atan2}(x_j, x_i).
$$

Because there are $\binom{4}{2} = 6$ unique pairs of axes, your 4D state naturally contains exactly **six angular amounts**:

$$
\theta_{12},\; \theta_{13},\; \theta_{14},\; \theta_{23},\; \theta_{24},\; \theta_{34}.
$$

---

### Grouping the 6 Angles into 3 Spheres

To visualize 6 numbers at once, we pair them up into three independent pairs:

| Sphere | Pair of Angles | Meaning |
| :---: | :--- | :--- |
| **Sphere 1** | $(\theta_{12},\; \theta_{34})$ | Relative orientation within the $(1,2)$ plane, combined with the $(3,4)$ plane. |
| **Sphere 2** | $(\theta_{13},\; \theta_{24})$ | Cross‑orientation between the $(1,3)$ and $(2,4)$ planes. |
| **Sphere 3** | $(\theta_{14},\; \theta_{23})$ | Cross‑orientation between the $(1,4)$ and $(2,3)$ planes. |

**Why a sphere for each pair?**  
A 2‑sphere ($S^2$) is the natural home for two angles: treat the first angle as **latitude** ($-\pi/2$ to $\pi/2$) and the second as **longitude** ($0$ to $2\pi$).  
Therefore, at any instant, your 4D state is mapped to **three distinct points**, one on each sphere.

---

### What are the "Two Circles" on each Sphere?

On a standard sphere, there are two fundamental families of circles:

- **Latitude circles** (parallels) – where the first angle is fixed, and the second varies.
- **Longitude circles** (meridians) – where the second angle is fixed, and the first varies.

These are the "two circles" per sphere. In your visualization:

- The **latitude circles** represent keeping one pairwise angle constant (e.g., $\theta_{12} = \text{const}$).
- The **longitude circles** represent keeping the other pairwise angle constant (e.g., $\theta_{34} = \text{const}$).

So, *"three spheres, each with two circles"* is just a crisp way of saying: **"We use three spherical coordinate grids to display our six independent angular observables."**

---

### What Does the Swap Coupling Do to This Picture?

This is where the visualization becomes powerful.

- **Without coupling** ($s_k = 0$): Each of the three points moves smoothly along some deterministic curve on its respective sphere. The spheres evolve independently.

- **With a swap** (e.g., swapping components $x_1$ and $x_2$):
  - The angle $\theta_{12}$ flips sign or shifts by $\pi$ (a reflection across the equator on Sphere 1).
  - The cross‑angles $\theta_{13}$ and $\theta_{23}$ (and by symmetry, others) also get reflected or rotated on Spheres 2 and 3.
  
  In visual terms: **a swap is a sudden, deterministic jump or reflection of the three points across specific great circles** on their respective spheres.

- **With strong coupling** (many $1$'s in your binary schedule): The points on all three spheres will "jitter" or "flip" frequently. The trajectory on each sphere will look like a smooth curve interrupted by mirror‑reflections.

---

### The Big Conceptual Payoff

Instead of trying to imagine a twisting 4D path, you watch **three moving points on three familiar 3‑D objects (spheres)**. 

- The **latitude** and **longitude** of each point tell you, at a glance, the phase relationships between the four coordinates.
- The **frequency of flips** (controlled by your coupling coefficient $C_N$) is directly visible as the rate at which these points bounce across the grid lines (the two circles) of each sphere.
- Most importantly, because the swap operator acts globally, you will see **synchronized reflections** across the three spheres—exposing the hidden cross‑dimensional symmetry that is otherwise invisible in a plain 4D scatter plot.

In short: **the idea is to use three 2‑spheres as "phase portraits" for the three mutually orthogonal complex planes embedded in your 4D space, turning an abstract vector‑field swap into a concrete ballet of reflected points on spherical grids.**
