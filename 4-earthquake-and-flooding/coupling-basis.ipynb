# Coupled Earthquake and Flood Equations

Two classical equations, each defined on an $N$-dimensional space, can be combined into the same $N$-dimensional framework through **coupling**.  
For a function $F(x, y)$, coupling means that the components of the gradient  
$$
\nabla F \;=\; \left( \frac{\partial F}{\partial x},\; \frac{\partial F}{\partial y} \right)
$$
are treated separately as the partial derivatives $\frac{\partial F}{\partial x}$ and $\frac{\partial F}{\partial y}$, rather than being summed together.  

The incorrect “nominal” approach would sum the partial derivatives:
$$
\frac{\partial F}{\partial x} + \frac{\partial F}{\partial y}
\quad\text{or, in $n$ dimensions,}\quad
\sum_{i=1}^{n} \frac{\partial F}{\partial x_i}.
$$
This sum is **not** the gradient; it is a directional derivative along the vector $(1,1,\dots,1)$, which loses the directional information essential for coupled systems. Keeping the gradient components separate allows each partial derivative to interact with the corresponding components of other equations, enabling a true coupled formulation.



# Continuous Coupling via a Smooth Mixing Function $\epsilon(\mathbf{x})$

## 1. Recalling the coupled gradient operator

We have two scalar fields on a domain $\Omega\subset\mathbb{R}^d$ ($d=1,2,3$):

$$
\phi,\;\psi \;:\; \Omega \longrightarrow \mathbb{R},
$$

assumed sufficiently smooth. Their individual gradients are $\nabla\phi$, $\nabla\psi$.  
We introduce a **coupling function** $\epsilon : \Omega \to \mathbb{R}$ and define the **coupled gradients** as

$$
\begin{aligned}
\mathbf{G}_\phi(\mathbf{x}) &:= \bigl(1-\epsilon(\mathbf{x})\bigr)\,\nabla\phi(\mathbf{x}) + \epsilon(\mathbf{x})\,\nabla\psi(\mathbf{x}),\\[4pt]
\mathbf{G}_\psi(\mathbf{x}) &:= \epsilon(\mathbf{x})\,\nabla\phi(\mathbf{x}) + \bigl(1-\epsilon(\mathbf{x})\bigr)\,\nabla\psi(\mathbf{x}).
\end{aligned}
\tag{1}
$$

At points where $\epsilon=0$ the fields decouple; where $\epsilon=1$ the gradients are completely swapped (the discrete limit).  
For a **continuous** coupling we require $0\le\epsilon(\mathbf{x})\le 1$, and to avoid degeneracy we will later see that a stricter bound is necessary.

## 2. Prototype PDE system

The coupled gradients appear in the equations of motion, for instance as a diffusive coupling:

$$
\begin{aligned}
\frac{\partial \phi}{\partial t} &= \nabla\cdot\bigl( \mathbf{G}_\phi \bigr) + f_\phi(\phi,\psi),\\[4pt]
\frac{\partial \psi}{\partial t} &= \nabla\cdot\bigl( \mathbf{G}_\psi \bigr) + f_\psi(\phi,\psi),
\end{aligned}
\tag{2}
$$

with suitable boundary conditions (e.g. homogeneous Dirichlet or Neumann) and smooth source terms $f_\phi,f_\psi$.  
We will analyse the linear (or semilinear) case where $\epsilon(\mathbf{x})$ is **prescribed** and does not depend on the solution.

The spatial operator for the pair $\mathbf{u} := (\phi,\psi)^T$ can be written as

$$
\mathcal{L}\mathbf{u} := 
\begin{pmatrix}
\nabla\cdot\bigl((1-\epsilon)\nabla\phi + \epsilon\nabla\psi\bigr) \\[4pt]
\nabla\cdot\bigl(\epsilon\nabla\phi + (1-\epsilon)\nabla\psi\bigr)
\end{pmatrix}
= \nabla\cdot\bigl( \mathbf{M}(\mathbf{x})\,\nabla\mathbf{u} \bigr),
\tag{3}
$$

where $\mathbf{M}(\mathbf{x})$ acts on the **vector** $\nabla\mathbf{u}$ (with each spatial component treated separately).  
Concretely, for each spatial direction $i=1,\dots,d$, the matrix coupling the derivatives $(\partial_i\phi,\partial_i\psi)^T$ is

$$
M(\mathbf{x}) = 
\begin{pmatrix}
1-\epsilon(\mathbf{x}) & \epsilon(\mathbf{x})\\[4pt]
\epsilon(\mathbf{x}) & 1-\epsilon(\mathbf{x})
\end{pmatrix}.
\tag{4}
$$

This symmetric matrix has eigenvalues $\lambda_1=1$ and $\lambda_2=1-2\epsilon(\mathbf{x})$.

## 3. Fundamental rule: parabolicity and well‑posedness

For the system (2) to be uniformly parabolic (and therefore admit a unique, stable solution), the matrix $M(\mathbf{x})$ must be **uniformly positive definite**:
there exists a constant $\nu>0$ such that

$$
\boldsymbol{\xi}^T M(\mathbf{x})\,\boldsymbol{\xi} \ge \nu\,|\boldsymbol{\xi}|^2
\qquad \forall\,\boldsymbol{\xi}\in\mathbb{R}^2,\; \forall\mathbf{x}\in\Omega.
\tag{5}
$$

Since the smallest eigenvalue is $\lambda_{\min}=1-2\epsilon$, we obtain the **strict upper bound**

$$
\boxed{\;\sup_{\mathbf{x}\in\Omega}\;\epsilon(\mathbf{x}) \;<\; \frac{1}{2}\;}
\quad\text{and}\quad
\boxed{\;\epsilon(\mathbf{x})\ge 0\; \text{(non‑negative coupling)}.}
\tag{Rule 1}
$$

If $\epsilon\ge\frac12$ somewhere, the operator becomes degenerate (one eigenvalue zero), which can lead to loss of uniqueness or to blow‑up in finite time. Therefore **Rule 1** is the first necessary condition for a well‑posed linear coupling.

**Proof sketch of Rule 1:**  
From (4) we have $\xi^T M\xi = (1-\epsilon)(\xi_1^2+\xi_2^2) + 2\epsilon\xi_1\xi_2 = (1-\epsilon)\|\xi\|^2 + \epsilon(\xi_1+\xi_2)^2 \ge (1-2\epsilon)\|\xi\|^2$ (using $(\xi_1+\xi_2)^2 \le 2\|\xi\|^2$). Hence $\nu = 1-2\sup\epsilon$, and positivity requires $\sup\epsilon<1/2$. This bound is sharp.

## 4. Smoothness requirements for $\epsilon$

Standard parabolic theory (Ladyzhenskaya–Solonnikov–Uraltseva) requires that the diffusion matrix $M_{ij}$ be at least Lipschitz continuous in space. Since $M$ depends linearly on $\epsilon$, it is sufficient to demand

$$
\boxed{\;\epsilon \in C^{0,1}(\overline{\Omega})\quad\text{(Lipschitz continuous)} \;.}
\tag{Rule 2}
$$

Often we may even take $\epsilon\in C^1(\overline{\Omega})$ to simplify energy estimates.  
If the domain has corners or $\epsilon$ is only piecewise smooth, additional compatibility conditions are needed, but we will work with a globally smooth $\epsilon$.

**Justification:** Lipschitz continuity guarantees that the weak divergence form $\nabla\cdot(M\nabla\mathbf{u})$ generates a strongly elliptic operator with $L^\infty$ coefficients, for which the $H^1$‑theory applies.

## 5. Existence and uniqueness theorem (linear, prescribed $\epsilon$)

**Theorem 1 (Well‑posedness)**  
Assume $\Omega\subset\mathbb{R}^d$ is a bounded Lipschitz domain. Let $\epsilon\in C^{0,1}(\overline{\Omega})$ satisfy $0\le\epsilon(\mathbf{x})\le \epsilon_{\max}<\frac12$.  
Consider the initial‑boundary value problem for (2) with, for simplicity, homogeneous Dirichlet boundary conditions and initial data $\mathbf{u}_0\in (L^2(\Omega))^2$.  
Then there exists a unique weak solution

$$
\mathbf{u}\in L^2(0,T; H^1_0(\Omega)^2)\cap C([0,T]; (L^2(\Omega))^2), \qquad \frac{d\mathbf{u}}{dt}\in L^2(0,T; H^{-1}(\Omega)^2).
$$

Moreover, the solution depends continuously on the data.

**Proof (energy method)**  
Define the bilinear form

$$
a(t;\mathbf{u},\mathbf{v}) = \int_\Omega \nabla\mathbf{v}^T : M(\mathbf{x})\,\nabla\mathbf{u}\,d\mathbf{x}
= \int_\Omega \Bigl[ (1-\epsilon)\nabla\phi\cdot\nabla\varphi + (1-\epsilon)\nabla\psi\cdot\nabla\theta + \epsilon(\nabla\psi\cdot\nabla\varphi + \nabla\phi\cdot\nabla\theta) \Bigr] d\mathbf{x},
$$

with $\mathbf{v}=(\varphi,\theta)$. Symmetry $a(\mathbf{u},\mathbf{v})=a(\mathbf{v},\mathbf{u})$ follows from $M^T=M$.  
Because $0\le\epsilon\le\epsilon_{\max}<\frac12$, the smallest eigenvalue of $M$ is $1-2\epsilon\ge 1-2\epsilon_{\max}=:\nu>0$. Consequently

$$
a(t;\mathbf{u},\mathbf{u}) \ge \nu \int_\Omega |\nabla\mathbf{u}|^2\,d\mathbf{x} = \nu\,\|\nabla\mathbf{u}\|_{L^2}^2.
\tag{6}
$$

Using Poincaré’s inequality we obtain coercivity in $H^1_0(\Omega)^2$:

$$
a(t;\mathbf{u},\mathbf{u}) \ge c\,\|\mathbf{u}\|_{H^1}^2.
\tag{7}
$$

The boundedness $|a(t;\mathbf{u},\mathbf{v})| \le C\|\mathbf{u}\|_{H^1}\|\mathbf{v}\|_{H^1}$ is obvious because $\epsilon$ is bounded.  
The linear source terms can be treated as perturbations. By standard Galerkin approximation and uniform energy estimates (multiply the PDE by $\mathbf{u}$ and integrate), we obtain global existence and uniqueness (e.g., Evans, *PDE*, §7.1). ∎

Thus **Rule 1** and **Rule 2** together guarantee a mathematically sound linear coupling.

## 6. Selecting $\epsilon$ for coordinate‑independent coupling

A pre‑defined function $\epsilon(\mathbf{x})$ breaks translation invariance. For many physical models we prefer $\epsilon$ to be **intrinsic** to the fields, e.g.

$$
\epsilon(\mathbf{x}) = \mathcal{E}\bigl( \|\nabla\phi(\mathbf{x})-\nabla\psi(\mathbf{x})\| \bigr),
\tag{8}
$$

where $\mathcal{E}:[0,\infty)\to[0,\tfrac12-\delta]$ is a smooth, bounded function.  
This makes the coupling **non‑linear**, as $\epsilon$ depends on the solution itself. To extend Theorem 1 we need:

- Lipschitz continuity of $\mathcal{E}$ so that $| \mathcal{E}(r_1)-\mathcal{E}(r_2)|\le L\,|r_1-r_2|$.
- A global bound $\mathcal{E}(r)\le \epsilon_{\max}<\frac12$ for all $r$.
- The gradient of the solution must then be controlled via a fixed‑point argument or by proving that the non‑linear term remains small for short times.

A typical result would be:

**Theorem 2 (Local well‑posedness for non‑linear coupling)**  
Under the same domain and data assumptions, if $\mathcal{E}$ is Lipschitz with $\sup \mathcal{E}<\frac12$, then there exists a unique local‑in‑time weak solution to (2) with $\epsilon$ given by (8).

*Proof idea:* Treat $\epsilon$ as a fixed function in the bilinear form, then use a contraction mapping on a short time interval using the energy estimates and the Lipschitz property of $\mathcal{E}$ to control the difference between two iterations.

**Rule 3 (for adaptive $\epsilon$):**  
Choose $\mathcal{E}\in C^{0,1}([0,\infty))$ with $\sup \mathcal{E} \le \epsilon_{\max}<\frac12$ and a bounded Lipschitz constant. Additionally, to ensure the gradient $\nabla\epsilon$ remains bounded in the energy estimates (when integrating by parts), one may need $\mathcal{E}\in C^1$ with bounded derivative, but this can be relaxed by using the chain rule and the fact that $\nabla\mathbf{u}\in L^2$ almost everywhere.

## 7. Summary: Rules for choosing $\epsilon(\mathbf{x})$

| Rule | Condition | Reason |
|------|-----------|--------|
| **R1** | $0 \le \epsilon(\mathbf{x}) \le \epsilon_{\max} < \frac12$ everywhere | Ensures uniform parabolicity (strict positive definiteness of $M$). |
| **R2** | $\epsilon \in C^{0,1}(\overline{\Omega})$ (Lipschitz) | Guarantees the PDE coefficients are regular enough for $H^1$‑theory. |
| **R3** | (Adaptive case) $\mathcal{E} \in C^{0,1}([0,\infty))$ with $\sup\mathcal{E}<\frac12$ | Allows well‑posedness of the resulting non‑linear system. |
| **R4** | (Optional) $\epsilon$ has compact support inside $\Omega$ | Avoids coupling at boundaries, simplifying boundary condition compatibility. |
| **R5** | (Optional) $\epsilon$ is radial / symmetric if the physics demands it | Maintains physical symmetries of the uncoupled problem. |

These rules provide a firm foundation for constructing continuous coupled earthquake‑flood models using gradient mixing.  

## 8. Conclusion and outlook

We have translated the discrete “swap” into a smooth operator parametrised by a scalar function $\epsilon(\mathbf{x})$, and derived necessary and sufficient conditions for the resulting PDE system to be well‑posed.  
The key finding is that $\epsilon$ must be **bounded strictly below $\frac12$**, a constraint that distinguishes a stable partial mixing from a degenerate full swap.  

Future work will apply these rules to specific earthquake‑flood coupling scenarios, selecting $\epsilon$ based on geophysical features (fault lines, river channels) and verifying numerically that the continuous model indeed captures the intended physical exchange.

# Continuous Coupling

The above equation is applicable for discrete functions with fixed intervals such as $\mathbb{Z}_n$. A more advanced definition is required to correctly apply coupling to continuous functions.

## 1. From discrete swap to continuous hybrid system

We start from the original discrete operation at a finite set of points $\{\mathbf{p}_k\}_{k=1}^{N_c}\subset\Omega\subset\mathbb{R}^d$. For two scalar fields $\phi,\psi:\Omega\to\mathbb{R}$, the **swap** at time $t$ (or as an instantaneous update) is

$$
\begin{pmatrix}
\nabla\phi(\mathbf{p}_k, t^+)\\[2pt]
\nabla\psi(\mathbf{p}_k, t^+)
\end{pmatrix}
\;=\;
\begin{pmatrix}
0 & I\\[2pt]
I & 0
\end{pmatrix}
\begin{pmatrix}
\nabla\phi(\mathbf{p}_k, t^-)\\[2pt]
\nabla\psi(\mathbf{p}_k, t^-)
\end{pmatrix}
\;=\;
\begin{pmatrix}
\nabla\psi(\mathbf{p}_k, t^-)\\[2pt]
\nabla\phi(\mathbf{p}_k, t^-)
\end{pmatrix}.
\tag{1}
$$

This exchanges the gradient vectors at those points while leaving the rest of the fields unchanged.  
Simply taking the limit as the set of points becomes dense would force the gradients to be equal almost everywhere—a trivial coupling.  
The correct continuous analogue is obtained by **keeping the exchange operation but applying it only on a lower‑dimensional set** and, in a time‑dependent setting, at discrete instants. This leads to a **hybrid PDE with impulsive reset**.

## 2. Geometry and function spaces

Let $\Omega\subset\mathbb{R}^d$ ($d\ge1$) be a bounded Lipschitz domain.  
Choose a closed set $\mathcal{C}\subset\overline{\Omega}$ of **Hausdorff dimension $d-1$** (e.g. a smooth curve in 2D, a surface in 3D).  
We assume $\mathcal{C}$ is sufficiently regular so that **trace operators** for gradients are well defined; a convenient choice is that $\mathcal{C}$ is a finite union of $C^{1,1}$ manifolds.

The state of the system at any time is a pair $(\phi,\psi)$ belonging to

$$
\mathbf{V} := H^1(\Omega) \times H^1(\Omega).
$$

For functions in $H^1(\Omega)$, the gradient $\nabla u$ is in $L^2(\Omega)^d$.  
If $\mathcal{C}$ is a Lipschitz hypersurface, the **trace of the gradient** on $\mathcal{C}$ is not defined in general. However, we can still make sense of the exchange by treating the coupling as a **jump condition** or by using a **dual formulation** with surface delta‑like forces.  
For rigorous analysis, it is often simpler to work with fields that are smooth enough to have pointwise gradient values on $\mathcal{C}$; this can be achieved by considering the PDE in a mild sense and assuming initial data in $H^2(\Omega)$.

Nevertheless, to present the core idea we temporarily assume that the fields are sufficiently regular and that the traces $\nabla\phi|_{\mathcal{C}}$ and $\nabla\psi|_{\mathcal{C}}$ exist (e.g., as elements of $L^2(\mathcal{C})^d$). We will then comment on the general case.

## 3. Impulsive reset operator $\mathcal{R}$

Define the **reset operator** $\mathcal{R} : \mathbf{V} \to \mathbf{V}$ that swaps the gradient traces on $\mathcal{C}$ and leaves the fields unchanged elsewhere. Formally,

$$
\mathcal{R}\begin{pmatrix} \phi \\ \psi \end{pmatrix}
\;=\;
\begin{pmatrix} \tilde\phi \\[2pt] \tilde\psi \end{pmatrix},
$$

where $\tilde\phi$ and $\tilde\psi$ are the unique functions in $H^1(\Omega)$ that satisfy

- $\tilde\phi = \phi$ and $\tilde\psi = \psi$ almost everywhere in $\Omega$,
- the gradient traces on $\mathcal{C}$ are exchanged:
  $$
  \gamma(\nabla\tilde\phi) = \gamma(\nabla\psi), \qquad
  \gamma(\nabla\tilde\psi) = \gamma(\nabla\phi),
  $$
  where $\gamma$ denotes the appropriate trace operator.

Because the bulk values are unchanged, the new functions remain in $H^1(\Omega)$ (one may check that the swapped gradient traces can be lifted to an $H^1$ extension consistent with the interior values; this is a nontrivial but solvable problem for sufficiently regular $\mathcal{C}$).  
In practical terms, $\mathcal{R}$ modifies the fields only in an infinitesimal neighbourhood of $\mathcal{C}$ so as to achieve the desired gradient traces.

## 4. Hybrid PDE formulation

We now embed this reset into a time‑evolution model. Suppose the uncoupled dynamics are given by a (possibly nonlinear) parabolic system:

$$
\partial_t \begin{pmatrix} \phi \\ \psi \end{pmatrix}
= \mathbf{A}\begin{pmatrix} \phi \\ \psi \end{pmatrix} + \mathbf{f}(t),
\qquad
\mathbf{A} = \begin{pmatrix} \nabla\cdot(\mathbf{D}_\phi\nabla) & 0 \\ 0 & \nabla\cdot(\mathbf{D}_\psi\nabla) \end{pmatrix},
\tag{2}
$$

with appropriate boundary conditions.  
We choose a sequence of **coupling times** $\{t_k\}_{k\in\mathbb{N}}$ with $0 < t_1 < t_2 < \dots$ (the impulsive instants) and define the evolution as follows:

- For $t \in (t_k, t_{k+1})$, the fields satisfy (2) (no coupling).
- At $t = t_k$, we apply the reset:
  $$
  \begin{pmatrix} \phi(t_k^+,\cdot) \\[2pt] \psi(t_k^+,\cdot) \end{pmatrix}
  \;=\;
  \mathcal{R}\begin{pmatrix} \phi(t_k^-,\cdot) \\[2pt] \psi(t_k^-,\cdot) \end{pmatrix}.
  \tag{3}
  $$

This is a **hybrid (impulsive) PDE**. It perfectly captures the discrete swap extended to a continuum of coupling locations: at the chosen instants, the gradient vectors of the two fields are exchanged pointwise on $\mathcal{C}$, just as they were exchanged at the isolated points $\mathbf{p}_k$ in the discrete version.

**Remark.** If the coupling must be continuous in time, one can replace the impulsive resets by a **continuous relaxation**:
$$
\frac{d}{dt}\begin{pmatrix} \phi \\ \psi \end{pmatrix}
= \mathbf{A}\begin{pmatrix} \phi \\ \psi \end{pmatrix}
+ \frac{1}{\tau}\Bigl(\mathcal{R} - \mathbf{I}\Bigr)\begin{pmatrix} \phi \\ \psi \end{pmatrix},
$$
which exponentially drives the gradient traces on $\mathcal{C}$ toward their swapped values with time constant $\tau$. This avoids discontinuities but is a regularisation, not the exact swap.

## 5. Well‑posedness theorem (sketch)

To guarantee that the hybrid system (2)–(3) makes sense, we need:

- The flow of the continuous part (2) exists and maps $\mathbf{V}$ into itself (or a more regular space where the trace of the gradient is defined).
- The reset map $\mathcal{R}$ is well defined and preserves the regularity required for the PDE in the next interval.

**Theorem 1 (Local well‑posedness of the impulsive coupling).**  
Assume:

1. $\mathbf{D}_\phi,\mathbf{D}_\psi$ are uniformly positive definite and Lipschitz.
2. The coupling set $\mathcal{C}$ is a closed $C^{2}$ hypersurface, and $\Omega$ is such that the trace operator $\gamma : H^{3/2}(\Omega) \to H^{1}(\mathcal{C})$ is surjective.  
3. The initial data $(\phi_0,\psi_0) \in H^2(\Omega)^2$ are compatible with the boundary conditions.
4. The sequence of coupling times satisfies $0 < t_1 < t_2 < \dots$ with $\inf_k (t_{k+1} - t_k) > 0$.

Then there exists a unique solution $(\phi,\psi)$ defined for all $t \ge 0$, which is piecewise $H^2$ in space and $C^1$ in time within each interval, and satisfies the reset conditions (3) at each $t_k$.

*Proof outline.*  
Between resets, standard parabolic theory gives a unique solution that stays in $H^2$ (since the diffusion coefficients are smooth and the domain is regular). The trace of the gradient on $\mathcal{C}$ is then in $H^{1/2}(\mathcal{C})$, which embeds into $L^2(\mathcal{C})$.  
The reset swaps these traces. The crucial point is to reconstruct $H^2$ functions on $\Omega$ having prescribed gradient traces on $\mathcal{C}$ and coinciding with the previous field outside an arbitrarily small neighbourhood of $\mathcal{C}$. This is a **lifting problem**; it can be solved using a right‑inverse of the trace operator for the gradient. Because $\mathcal{C}$ has codimension 1, such a lifting exists and can be made smooth and with compact support around $\mathcal{C}$. Hence $\mathcal{R}$ maps $H^2(\Omega)^2$ continuously into itself.  
By induction, the solution stays in $H^2$ at the beginning of each interval, so the process can be continued. Uniqueness follows from the deterministic nature of the flow and the reset map.

**Remark on general function spaces.**  
If one wishes to work in the natural energy space $H^1(\Omega)$, a direct gradient trace is not available. One can instead interpret the reset as a **singular forcing** at the interface. The PDE is then written as

$$
\partial_t \begin{pmatrix} \phi \\ \psi \end{pmatrix}
= \mathbf{A}\begin{pmatrix} \phi \\ \psi \end{pmatrix}
+ \delta_{\mathcal{C}} \cdot \mathbf{J},
$$

where $\mathbf{J}$ is a surface current chosen so that the jump in the gradient across $\mathcal{C}$ corresponds exactly to the swap. This leads to a transmission problem that is also well‑posed. The detailed analysis requires interface Sobolev spaces and is beyond the present scope.

## 6. Connection to the discrete $\mathbb{Z}_n$ version

To recover the discrete coupling on a grid $\mathbb{Z}_n^d$ (or a finite set of points), we specialise $\mathcal{C}$ to be exactly that finite set. The trace operator on a set of isolated points reduces to point evaluation, which is well defined for $H^{s}(\mathbb{R}^d)$ with $s > d/2$ (a condition that holds for $H^2$ when $d\le3$). The reset operator then becomes exactly the finite‑dimensional swap (1). Thus the hybrid PDE framework seamlessly unifies the discrete and continuous coupling concepts.

## 7. Summary

We have defined a rigorous continuous coupling that mirrors the discrete gradient swap by:

- Restricting the exchange to a lower‑dimensional subset $\mathcal{C}$.
- Applying the exchange at discrete times as an impulsive reset.
- Formulating the reset via a trace swap operator $\mathcal{R}$ that preserves the necessary spatial regularity for the subsequent PDE evolution.

This approach avoids the trivialisation that would occur with a pointwise equality condition and provides a mathematically sound foundation for further analysis and numerical simulation.

# Reducing Lumped Earthquake and Flood Equations to Wave Equations for Gradient Coupling

The original coupling scheme swaps (or mixes) the **gradients** of two scalar fields  
$\phi(\mathbf{x},t)$ and $\psi(\mathbf{x},t)$.  
To use that scheme, we must first translate the classical lumped formulas  

$$
M_0 = \mu\,\bar{u}\,S \qquad\text{(seismic moment)}
$$  
$$
Q = C\,A^{\,n} \qquad\text{(flood discharge)}
$$

into continuous field descriptions whose **gradients** are the physical quantities that drive the dynamics.

---

## 1. Earthquake Field: from Seismic Moment to a Wave Equation

### 1.1 Slip Potential

Seismic moment is the product of shear modulus $\mu$, average fault slip $\bar{u}$, and fault area $S$.  
Slip is a **displacement discontinuity** across a fault surface $\Gamma$.  
To represent it by a continuous scalar field, introduce a **slip potential** $\phi(\mathbf{x},t)$ such that the local slip magnitude (or the component of slip in the dominant direction) is proportional to the gradient of $\phi$ normal to the fault:

$$
u(\mathbf{x},t) \;\propto\; \nabla\phi(\mathbf{x},t)\cdot\mathbf{n}, \qquad \mathbf{x}\in\Gamma,
$$

where $\mathbf{n}$ is the unit normal to the fault.  
In a simplified 1D fault model we may simply set $u = L\,\partial_x\phi$.

The seismic moment **density** (moment per unit fault area) is $m_0 = \mu\,u$, so the total moment becomes

$$
M_0(t) = \int_\Gamma \mu(\mathbf{x})\,u(\mathbf{x},t)\,dS \;\propto\; \int_\Gamma \mu\,|\nabla\phi|\,dS.
\tag{1}
$$

Thus the gradient $\nabla\phi$ directly encodes the slip that generates seismic waves.

### 1.2 Wave Equation for $\phi$

For a mode‑III (anti‑plane) rupture or a simplified scalar model, the displacement potential satisfies the scalar wave equation

$$
\frac{\partial^2 \phi}{\partial t^2} = c_s^2\,\Delta\phi + F_\phi(\nabla\phi,\dots),
\tag{2}
$$

where $c_s$ is the shear‑wave speed and $F_\phi$ represents source terms (e.g., friction laws expressed in terms of $\nabla\phi$).  
Equation (2) is the **earthquake wave field** we will couple to the flood field.

### 1.3 Summary of Reduction ($M_0 \to \phi$)

1. Postulate a scalar potential $\phi$ whose gradient gives the slip.  
2. The wave equation (2) governs its evolution.  
3. The seismic moment $M_0$ is obtained diagnostically from $\nabla\phi$ using (1).

---

## 2. Flood Field: from Discharge Rating to a Wave Equation

### 2.1 Water Depth as the Primary Variable

The empirical rating curve $Q = C A^n$ is not a dynamical law; it is derived from a combination of flow resistance (Manning, Chézy) and channel geometry.  
The underlying physics is captured by the **shallow‑water equations**. In one dimension, with $A(x,t)$ the cross‑sectional area and $h(x,t)$ the water surface elevation, the mass and momentum equations are

$$
\frac{\partial A}{\partial t} + \frac{\partial Q}{\partial x} = 0,
\tag{3}
$$
$$
\frac{\partial Q}{\partial t} + \frac{\partial}{\partial x}\!\left(\frac{Q^2}{A}\right) + gA\frac{\partial h}{\partial x} = gA(S_0 - S_f).
\tag{4}
$$

Here $S_0$ is the bed slope and $S_f$ the friction slope. The discharge $Q$ is related to the water‑surface slope $\partial h/\partial x$ through a resistance law, e.g. Manning’s equation:

$$
Q = \frac{1}{n} A\,R^{2/3}\,\sqrt{\,|S_f|\,},
\qquad
S_f \approx \frac{\partial h}{\partial x} + \text{(other terms)}.
$$

Hence the gradient $\partial h/\partial x$ **directly controls the discharge**.

### 2.2 Diffusive Wave Approximation

To obtain a single scalar PDE for $h$ (which we rename $\psi$), we often neglect inertial terms (the first two terms in (4)) and obtain the **diffusive wave equation**

$$
\frac{\partial \psi}{\partial t} = \frac{\partial}{\partial x}\!\left( D(\psi)\,\frac{\partial \psi}{\partial x} \right) + \text{sources/sinks},
\tag{5}
$$

where $D(\psi)$ is a nonlinear diffusion coefficient derived from the resistance law and channel geometry.  
Equation (5) is parabolic but can be supplemented with the full hyperbolic Saint‑Venant system if wave propagation effects are important.

### 2.3 Summary of Reduction ($Q = C A^n \to \psi$)

1. Let $\psi(\mathbf{x},t) = h(\mathbf{x},t)$ be the water surface elevation.  
2. The gradient $\nabla\psi$ is the water‑surface slope that determines flow.  
3. The governing PDE for $\psi$ is the diffusive wave (or shallow‑water) equation (5).  
4. Discharge $Q$ is a diagnostic output computed from $\nabla\psi$ and the cross‑sectional area (which depends on $\psi$).

---

## 3. Coupling the Two Wave Equations

Now both fields are described by continuous wave‑type equations:

- Earthquake: $\displaystyle \frac{\partial^2\phi}{\partial t^2} = c_s^2\Delta\phi + \dots$  
- Flood: $\displaystyle \frac{\partial\psi}{\partial t} = \nabla\!\cdot\!\bigl(D(\psi)\nabla\psi\bigr) + \dots$

The coupling is applied to the **gradients** of these fields.

### 3.1 Impulsive Swap (exact discrete coupling)

At a set of coupling points (or a lower‑dimensional set $\mathcal{C}$) and at coupling instants $t_k$, we exchange the gradient vectors:

$$
\begin{pmatrix}
\nabla\phi(\mathbf{x}, t_k^+) \\[2pt]
\nabla\psi(\mathbf{x}, t_k^+)
\end{pmatrix}
=
\begin{pmatrix}
\nabla\psi(\mathbf{x}, t_k^-) \\[2pt]
\nabla\phi(\mathbf{x}, t_k^-)
\end{pmatrix},
\qquad \mathbf{x}\in\mathcal{C}.
\tag{6}
$$

### 3.2 Continuous Smooth Blending

For a regularised continuous coupling, we define a smooth coupling function $\epsilon(\mathbf{x})$ and replace the gradients in the PDEs by the mixed vectors

$$
\mathbf{G}_\phi = \bigl(1-\epsilon(\mathbf{x})\bigr)\,\nabla\phi + \epsilon(\mathbf{x})\,\nabla\psi,
$$
$$
\mathbf{G}_\psi = \epsilon(\mathbf{x})\,\nabla\phi + \bigl(1-\epsilon(\mathbf{x})\bigr)\,\nabla\psi.
\tag{7}
$$

The coupled system then becomes

$$
\frac{\partial^2 \phi}{\partial t^2} = c_s^2\,\nabla\!\cdot\!\mathbf{G}_\phi + \dots,
\qquad
\frac{\partial \psi}{\partial t} = \nabla\!\cdot\!\bigl( D(\psi)\,\mathbf{G}_\psi \bigr) + \dots
\tag{8}
$$

(Note: In the earthquake equation we take the divergence of $\mathbf{G}_\phi$ to form a Laplacian‑like coupling; in the flood equation we multiply the mixed gradient by the nonlinear diffusion coefficient $D(\psi)$.)

### 3.3 Recovery of the Lumped Quantities

After solving the coupled field system, the traditional lumped measures can be calculated diagnostically:

- **Seismic moment:**  
  $$
  M_0(t) = \int_{\Gamma} \mu\,|\nabla\phi(\mathbf{x},t)|\,dS,
  $$
  or, if a scalar slip law is used, $M_0 = \mu L \int_{\Gamma} \partial_n\phi\,dS$.

- **Flood discharge at a cross‑section:**  
  $$
  Q(x,t) = \frac{1}{n}A(\psi)\,R(\psi)^{2/3}\,\sqrt{\,\left|\frac{\partial\psi}{\partial x}\right|\,},
  $$
  which reduces to the classical rating curve $Q = C A^n$ when the slope is constant or parameterized.

---

## 4. Summary of the Reduction

| Original lumped formula | Continuous field | Gradient meaning | Wave equation |
|--------------------------|------------------|------------------|---------------|
| $M_0 = \mu\bar{u}S$      | Slip potential $\phi$ | $\nabla\phi \propto$ slip | $\ddot\phi = c_s^2\Delta\phi$ |
| $Q = C A^n$              | Water depth $\psi$    | $\nabla\psi \propto$ water slope | $\dot\psi = \nabla\!\cdot\!(D(\psi)\nabla\psi)$ |

**Coupling:** exchange or mix $\nabla\phi$ and $\nabla\psi$ at fault–river intersections.  
The coupled system (8) now fully defines the interaction between earthquake rupture waves and flood propagation waves, while the classical $M_0$ and $Q$ formulas serve as diagnostic outputs.

This reduction is the essential step that allows the gradient‑swapping framework to be applied to realistic geophysical problems.
