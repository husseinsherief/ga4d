# Conventional Gradient Functions

The following is the definition of a conventional gradient function $\nabla F$.

## 1. Motivation: Keeping Gradient Components Separate

Consider a smooth scalar field $F(x, y)$ defined on an $N$-dimensional space. The gradient
$$
\begin{aligned}
\nabla F &= \left( \frac{\partial F}{\partial x},\; \frac{\partial F}{\partial y} \right) \\
         &= \left( \frac{f(x + \Delta x_i) - f(x - \Delta x_i)}{2 \Delta x_i}, \frac{f(y + \Delta y_i) - f(y - \Delta y_i)}{2 \Delta y_i} \right).
\end{aligned}
$$
is an ordered tuple of partial derivatives. In a coupled system of equations, these components must remain separate so that each partial derivative can interact with the corresponding component of another equation.

---

## 2. Theorem 1: Definition of the Gradient as a Covector Field

**Theorem.** Let $M$ be a smooth $n$-dimensional manifold and let $F: M \to \mathbb{R}$ be a smooth scalar field. The gradient $\nabla F$ is an element of the cotangent bundle $T^*M$, not a scalar. Its components are $\partial_i F$, but the object itself is not their sum.

**What this establishes.** The gradient is an ordered tuple of components bound to a basis. Summing the components discards the basis and destroys the covector structure required for geometric coupling.

---

### Proof of Theorem 1

The fundamental object describing the rate of change of $F$ is the exterior derivative $dF$. At any point $p \in M$, $dF$ is a linear map from the tangent space $T_pM$ to $\mathbb{R}$; by definition, it is a covector in the cotangent space $T^*_pM$.

Let $(x^1, x^2, \dots, x^n)$ be a local coordinate chart on $M$. The natural basis for $T^*_pM$ is $\{dx^1, dx^2, \dots, dx^n\}$, and $dF$ expands as
\begin{equation}
    dF = \sum_{i=1}^n \frac{\partial F}{\partial x^i}\, dx^i.
    \tag{1}
\end{equation}

Equation (1) shows that $dF$ (identified with $\nabla F$ in Euclidean space) is the tensor product of its components and the basis covectors. The components $\frac{\partial F}{\partial x^i}$ are fundamentally separate, each tied to its basis direction $dx^i$. Summing the components $\sum_{i=1}^n \frac{\partial F}{\partial x^i}$ discards the basis $\{dx^i\}$ and yields a scalar. A scalar cannot act as a linear map on tangent vectors; therefore, the component sum is a categorical error that destroys the covector nature required for geometric coupling. $\blacksquare$

---

## 3. Theorem 2: Directional Derivative Identity

**Theorem.** Let $F: \mathbb{R}^n \to \mathbb{R}$ be a differentiable scalar field. The scalar sum $\sum_{i=1}^n \frac{\partial F}{\partial x_i}$ is the directional derivative of $F$ along the vector $\mathbf{u} = (1,1,\dots,1)$, not the gradient itself.

**What this establishes.** The component sum is a single scalar value representing a one-dimensional slice of the function's behavior, entirely devoid of the $N$-dimensional vector information required to couple separate partial differential equations.

---

### Proof of Theorem 2

Let $\mathbf{v} \in \mathbb{R}^n$ be an arbitrary vector. The directional derivative of $F$ in the direction of $\mathbf{v}$ is given by the dot product
\begin{equation}
    D_{\mathbf{v}}F = \nabla F \cdot \mathbf{v}.
    \tag{2}
\end{equation}

Let $\mathbf{u} = (1, 1, \dots, 1)^T$. Substituting into (2) gives
\begin{equation}
    D_{\mathbf{u}}F = \nabla F \cdot \mathbf{u}.
    \tag{3}
\end{equation}

Expanding the dot product,
\begin{equation}
    D_{\mathbf{u}}F = \sum_{i=1}^n \frac{\partial F}{\partial x_i}\, u_i.
    \tag{4}
\end{equation}

Since $u_i = 1$ for all $i$, this simplifies to
\begin{equation}
    D_{\mathbf{u}}F = \sum_{i=1}^n \frac{\partial F}{\partial x_i}.
    \tag{5}
\end{equation}

Equation (5) proves that the component sum is identical to the directional derivative $D_{\mathbf{u}}F$ along $(1,1,\dots,1)$. It is a single scalar representing a one-dimensional slice of $F$, entirely devoid of the $N$-dimensional vector information required to couple separate partial differential equations. $\blacksquare$

## 4. Theorem 3: Tensor Transformation Law

**Theorem.** Let $x^i$ and $\tilde{x}^j$ be two coordinate systems on the same $N$-dimensional space. The component sum $S = \sum_{i=1}^n \frac{\partial F}{\partial x^i}$ does not transform as a scalar invariant under general coordinate changes; its value depends on the choice of coordinates.

**What this establishes.** Because the component sum is not a geometric invariant, summing the components irrecoverably destroys the directional information essential for coupled systems.

---

### Proof of Theorem 3

Let the components of the gradient in the original system be $V_i = \frac{\partial F}{\partial x^i}$.

First, consider the correct transformation. By the tensor transformation law for a covariant rank-1 tensor, the components in the new coordinate system, $\tilde{V}_j$, must be
\begin{equation}
    \tilde{V}_j = \sum_{i=1}^n \frac{\partial x^i}{\partial \tilde{x}^j}\, V_i.
    \tag{6}
\end{equation}
This transformation ensures the geometric object $\mathbf{V}$ remains invariant; the physical meaning is preserved because the components transform to compensate for the change in basis vectors.

Now, define the component sum in the original system as $S = \sum_{i=1}^n V_i$. If we calculate the equivalent sum in the new coordinate system, $\tilde{S}$, we substitute Equation (6):
\begin{equation}
    \tilde{S} = \sum_{j=1}^n \tilde{V}_j = \sum_{j=1}^n \left( \sum_{i=1}^n \frac{\partial x^i}{\partial \tilde{x}^j}\, V_i \right).
    \tag{7}
\end{equation}

Factoring out the components $V_i$, we obtain
\begin{equation}
    \tilde{S} = \sum_{i=1}^n V_i \left( \sum_{j=1}^n \frac{\partial x^i}{\partial \tilde{x}^j} \right).
    \tag{8}
\end{equation}

For $\tilde{S}$ to equal the original sum $S$ (i.e., for the sum to be a geometric invariant), the inner summation $\sum_{j=1}^n \frac{\partial x^i}{\partial \tilde{x}^j}$ must equal $1$ for all $i$. However, it is possible to construct $x^i$ and $\tilde{x}^j$ such that $\sum_{j=1}^n \frac{\partial x^i}{\partial \tilde{x}^j}$ is not equal to 1. For instance, by the inverse function theorem, the matrix $\left[ \frac{\partial x^i}{\partial \tilde{x}^j} \right]$ is the inverse Jacobian of the coordinate transformation. The sum of the elements in each row of an inverse Jacobian matrix is generally not equal to $1$ (except for the trivial identity transformation).

Because $\tilde{S} \neq S$ under a general coordinate rotation, the component sum is not a geometric invariant. Its value depends arbitrarily on the orientation of the observer's coordinate axes. By summing the components, the directional information is irrecoverably destroyed. $\blacksquare$
