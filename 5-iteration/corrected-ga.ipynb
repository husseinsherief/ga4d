# Corrected Coupled Flood–Earthquake Equations in Two 4D Spacetime Sectors

The flood and seismic fields live in **two different 4D Clifford algebras** sharing only the $(t_1,x_1)$ surface.

---

## 1. Spacetime & Geometric Algebra Setup

### Flood sector $\mathcal{F}$

$$
X_F=(t_1,x_1,t_2,x_2)
$$

Basis:

$$
\{\gamma_{t_1},\gamma_{x_1},\gamma_{t_2},\gamma_{x_2}\}
$$

Metric:

$$
\gamma_{t_i}^2=+1,\qquad \gamma_{x_i}^2=-1
$$

$$
\eta_F=\mathrm{diag}(+1,-1,+1,-1)
$$

Spacetime derivative:

$$
\nabla_F=\gamma^{t_1}\partial_{t_1}+\gamma^{x_1}\partial_{x_1}
+\gamma^{t_2}\partial_{t_2}+\gamma^{x_2}\partial_{x_2}
$$

Physical flood fields are independent of $t_2$:

$$
\partial_{t_2}h=\partial_{t_2}v_1=\partial_{t_2}v_2=0
$$

---

### Seismic sector $\mathcal{S}$

$$
X_S=(t_1,x_1,t_3,x_3)
$$

Basis:

$$
\{\gamma_{t_1},\gamma_{x_1},\gamma_{t_3},\gamma_{x_3}\}
$$

Metric:

$$
\eta_S=\mathrm{diag}(+1,-1,+1,-1)
$$

Spacetime derivative:

$$
\nabla_S=\gamma^{t_1}\partial_{t_1}+\gamma^{x_1}\partial_{x_1}
+\gamma^{t_3}\partial_{t_3}+\gamma^{x_3}\partial_{x_3}
$$

---

### Shared coupling surface

$$
\mathcal{C}=(t_1,x_1)
$$

The flood sector and seismic sector are coupled only through this common surface.

## 2. Multivector Fields

### Flood momentum multivector

$$
M=h\gamma_{t_1}+h v_1\gamma_{x_1}+h v_2\gamma_{x_2}
$$

where

$$
h=h(t_1,x_1,x_2),\qquad v_i=v_i(t_1,x_1,x_2)
$$

### Lagrange multiplier

$$
\Lambda=\Lambda(t_1,x_1,x_2)
$$

This is a scalar in the flood sector. Do not confuse it with the Lamé constant $\lambda$.

### Seismic displacement

$$
\mathbf{u}=u_1(t_1,t_3,x_1,x_3)\gamma_{x_1}
+u_3(t_1,t_3,x_1,x_3)\gamma_{x_3}
$$

## 3. Coupling Source

$$
S[\mathbf{u}]
=
-g h\,\frac{\partial z_b}{\partial x_1}\gamma_{t_1}
$$

where the bed elevation is obtained from the seismic vertical displacement on the shared surface:

$$
z_b(t_1,t_3,x_1)
=
z_{b0}(x_1)
-
u_3(t_1,t_3,x_1,x_3=0)
$$

Thus

$$
\frac{\partial z_b}{\partial t_1}
=
-\frac{\partial u_3}{\partial t_1},\qquad
\frac{\partial z_b}{\partial t_3}
=
-\frac{\partial u_3}{\partial t_3}
$$

## 4. Corrected Total Action

$$
I_{\text{total}}[M,\Lambda,\mathbf{u}]
=
\int_{\mathcal{M}_F}
\langle \Lambda(\nabla_F M-S[\mathbf{u}])\rangle_0\,d^4X_F
+
\int_{\mathcal{M}_S}
\mathcal{L}_{\text{seismic}}[\mathbf{u}]\,d^4X_S
$$

where

$$
\mathcal{L}_{\text{seismic}}
=
\frac12\rho(\partial_{t_1}\mathbf{u}+\partial_{t_3}\mathbf{u})^2
-\frac12(\lambda+2\mu)(\nabla_S\cdot\mathbf{u})^2
+\frac12\mu(\nabla_S\wedge\mathbf{u})^2
$$

$$
\boxed{

=
-\partial_{x_1}(\Lambda g h)\gamma_{x_3}

$$

## 6. Expanded Component Form

### (a) Flood equations

These evolve along the shared time $t_1$, with no $t_2$-dependence:

$$
\boxed{
\begin{aligned}
&\partial_{t_1}h
+\partial_{x_1}(h v_1)
+\partial_{x_2}(h v_2)
=0
\\[4pt]
&\partial_{t_1}(h v_1)
+\partial_{x_1}\left(h v_1^2+\frac12 g h^2\right)
+\partial_{x_2}(h v_1 v_2)
=
-g h\frac{\partial z_b}{\partial x_1}
\\[4pt]
&\partial_{t_1}(h v_2)
+\partial_{x_1}(h v_1 v_2)
+\partial_{x_2}\left(h v_2^2+\frac12 g h^2\right)
=
-g h\frac{\partial z_b}{\partial x_2}
\end{aligned}
}
$$

If the bed varies only in $x_1$, then

$$
\frac{\partial z_b}{\partial x_2}=0
$$

and the last equation reduces to

$$
\partial_{t_1}(h v_2)
+\partial_{x_1}(h v_1 v_2)
+\partial_{x_2}\left(h v_2^2+\frac12 g h^2\right)
=0
$$

---

### (b) Seismic wave equation

The seismic displacement $\mathbf{u}$ evolves along both $t_1$ and $t_3$. In component form:

$$
\boxed{
\begin{aligned}
&\rho(\partial_{t_1}+\partial_{t_3})^2 u_1
-
(\lambda+2\mu)\partial_{x_1}(\partial_{x_1}u_1+\partial_{x_3}u_3)
-
\mu(\partial_{x_1}^2+\partial_{x_3}^2)u_1
=0
\\[4pt]
&\rho(\partial_{t_1}+\partial_{t_3})^2 u_3
-
(\lambda+2\mu)\partial_{x_3}(\partial_{x_1}u_1+\partial_{x_3}u_3)
-
\mu(\partial_{x_1}^2+\partial_{x_3}^2)u_3
=
+\partial_{x_1}(\Lambda g h)
\end{aligned}
}
$$

The forcing appears only in the $u_3$-component because

$$
\mathbf{F}_{\text{flood}}
=
-\partial_{x_1}(\Lambda g h)\gamma_{x_3}
$$

### 7. Discrete‑Time Dynamical System (Explicit Euler)

Let the state at step $k$ be $\mathbf{x}_k \in \mathbb{R}^n$, and let the step size be $h > 0$.  
Given a fixed permutation matrix $P$ (e.g., swapping coordinates $i$ and $j$) and a deterministic binary schedule $\{s_k\}_{k=0}^{\infty} \subset \{0,1\}$, the Euler iteration is:

$$
\mathbf{x}_{k+1} = \mathbf{x}_k + h \cdot \mathbf{F}^{(s_k)}(\mathbf{x}_k),
$$

where

$$
\mathbf{F}^{(s_k)}(\mathbf{x}_k) = \big[(1 - s_k) I + s_k P\big] \mathbf{F}(\mathbf{x}_k).
$$

Expanding the bracket:

$$
\boxed{\mathbf{x}_{k+1} = \mathbf{x}_k + h \Big( (1 - s_k)\,\mathbf{F}(\mathbf{x}_k) + s_k \, P\mathbf{F}(\mathbf{x}_k) \Big)}
$$


### 8. Discrete Coupling Coefficient

Over $N$ iterations, the coupling coefficient exactly matches your empirical frequency definition:

$$
\boxed{C_N = \frac{1}{N} \sum_{k=0}^{N-1} s_k = \frac{\text{number of } 1\text{s in } S}{N}}
$$

This scalar lies in $[0,1]$. A schedule of all zeros gives $C_N = 0$ (uncoupled), all ones gives $C_N = 1$ (fully swapped at every step), and a mixture gives intermediate coupling strength.

## Seismic displacement $\mathbf{u}$

The seismic displacement lives in the sector with coordinates $(t_1, x_1, t_3, x_3)$:

$$
\mathbf{u} = u_1 \gamma_{x_1} + u_3 \gamma_{x_3}
$$

where

- $u_1 = u_1(t_1,t_3,x_1,x_3)$ is the **horizontal seismic displacement** along the $x_1$-axis.
- $u_3 = u_3(t_1,t_3,x_1,x_3)$ is the **vertical seismic displacement** along the $x_3$-axis.

There is **no $u_2$** in the seismic sector because the seismic coordinates do not include $x_2$. The two seismic spatial axes are $x_1$ and $x_3$.

The coupling uses the vertical component at the ground surface:

$$
z_b = z_{b0}(x_1) - u_3\big|_{x_3=0}
$$

So $u_3$ plays the role of the ground uplift/subsidence.

## Flood velocity components

The flood flow has two horizontal velocity components in the $(x_1,x_2)$ plane:

$$
v_1 = v_1(t_1,x_1,x_2)
$$

$$
v_2 = v_2(t_1,x_1,x_2)
$$

These are **depth-averaged horizontal velocities** in the shallow-water equations. They are not seismic displacements; they are flood velocities.

In the standard shallow-water flood model, $t_2$ is **not used as a dynamical time**. 

In the two-sector formulation, the flood sector is defined in $(t_1,x_1,t_2,x_2)$ to be a 4D spacetime with two time dimensions, but the actual flood physics only evolves along the shared time $t_1$. 

The flood momentum multivector is:

$$
M = h\gamma_{t_1} + h v_1 \gamma_{x_1} + h v_2 \gamma_{x_2}
$$

| $u_1$ | horizontal seismic displacement along $x_1$ |
| $u_3$ | vertical seismic displacement along $x_3$ |
| $v_1$ | flood horizontal velocity in $x_1$-direction |
| $v_2$ | flood horizontal velocity in $x_2$-direction |
| $h$ | flood water depth |
| $\Lambda$ | Lagrange multiplier / adjoint flood variable |
| $z_b$ | bed elevation, coupled to seismic vertical displacement |

There is no $u_2$ because the seismic model is 2D in space $(x_1,x_3)$, while the flood model is 2D in space $(x_1,x_2)$.

### Why $t_2$ is not in the flood equation

In the standard shallow-water flood model, $t_2$ is **not used as a dynamical time**. 

In the two-sector formulation, the flood sector is defined in $(t_1,x_1,t_2,x_2)$ to be a 4D spacetime with two time dimensions, but the actual flood physics only evolves along the shared time $t_1$. Therefore we impose:

$$
\partial_{t_2} h = \partial_{t_2} v_1 = \partial_{t_2} v_2 = 0
$$

This makes $t_2$ a **passive, non-evolving coordinate** in the flood sector. It is present only to match the geometric structure (a 4D spacetime with two time directions) so that the flood and seismic sectors can share the same $(t_1,x_1)$ surface while each retains its own extra time dimension.

If one wanted a truly two-time flood dynamics (e.g., a theoretical model with two independent time flows), the $\partial_{t_2}$ terms would appear in the flood equations. But for physical coupling with earthquakes, the flood is taken to be a standard shallow-water system that evolves only in $t_1$. Thus, $t_2$ is **formally defined but not used** in the equations.
