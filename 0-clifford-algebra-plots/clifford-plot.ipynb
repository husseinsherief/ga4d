```python
from clifford.g3c import *
from clifford.tools.g3c import *
from pyganja import GanjaScene, draw

# Create objects
P1 = up(random_euc_mv() * 0.1)
P2 = up(random_euc_mv() * 0.1)
P3 = up(random_euc_mv() * 0.1)
P4 = up(random_euc_mv() * 0.1)
P5 = up(random_euc_mv() * 0.1)

S = (P1 ^ P2 ^ P3 ^ P4 ^ P5).normal()
L = P1 ^ P2 ^ einf

# Build scene
sc = GanjaScene()
sc.add_objects([P1, P2, P3, P4, P5], color=(0,0,0))    # black
sc.add_objects([L], color=(0,0,255))    # blue
sc.add_objects([S], color=(255,0,0))    # red

# Open in a new browser tab (interactive, no notebook JS issues)
draw(sc, scale=0.5, sig=layout.sig, new_window=True)
```
